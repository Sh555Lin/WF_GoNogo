```python
import os
import cv2
import shutil
from pathlib import Path

def extract_sample_frames(src_root, dst_root, animals, frame_num=100):
    src_root = Path(src_root)
    dst_root = Path(dst_root)
    dst_root.mkdir(parents=True, exist_ok=True)

    for animal in animals:
        animal_path = src_root / animal
        if not animal_path.exists():
            print(f"[!] Animal folder not found: {animal_path}")
            continue
        
        for date_folder in sorted(animal_path.iterdir()):
            if not date_folder.is_dir():
                continue
            
          
            for video_file in date_folder.glob("*.avi"):
                try:
                    rel_path = video_file.relative_to(src_root)
                    dst_video_path = dst_root / rel_path
                    dst_video_path.parent.mkdir(parents=True, exist_ok=True)


                    cap = cv2.VideoCapture(str(video_file))
                    fps = cap.get(cv2.CAP_PROP_FPS)
                    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
                    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

                    fourcc = cv2.VideoWriter_fourcc(*'XVID')
                    out = cv2.VideoWriter(str(dst_video_path), fourcc, fps, (width, height))

                    count = 0
                    while count < frame_num:
                        ret, frame = cap.read()
                        if not ret:
                            break
                        out.write(frame)
                        count += 1

                    cap.release()
                    out.release()
                    print(f"[✓] Saved {count} frames → {dst_video_path}")


                    txt_file = video_file.with_suffix(".txt")
                    if txt_file.exists():
                        dst_txt_path = dst_video_path.with_suffix(".txt")
                        shutil.copy2(txt_file, dst_txt_path)
                        print(f"[✓] Copied timestamp file → {dst_txt_path}")
                    else:
                        print(f"[!] Timestamp file not found for {video_file.name}")

                except Exception as e:
                    print(f"[✗] Error processing {video_file}: {e}")

if __name__ == "__main__":
    src_root = "/home/lsh/Data_attention/Transfer learning/DATA_linshu"
    dst_root = "/home/lsh/Data_attention/Transfer learning/test"
    animals = ["A095"]  
    extract_sample_frames(src_root, dst_root, animals, frame_num=100)
```
